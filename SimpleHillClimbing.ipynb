

# Simple Hill Climbing

---




```python
# Simple Hill Climbing with Graph
# Function: f(x) = -(x - 3)^2 + 9

import matplotlib.pyplot as plt
import numpy as np


# Objective function
def objective_function(x):
    return -(x - 3) ** 2 + 9


# Simple Hill Climbing Algorithm
def simple_hill_climbing(start):
    current = start
    current_value = objective_function(current)

    # Store visited points for graph
    path = [(current, current_value)]

    print("Starting at x =", current, "f(x) =", current_value)

    while True:
        # Two neighbors: left and right
        neighbors = [current - 1, current + 1]
        found_better = False

        # Check neighbors one by one
        for neighbor in neighbors:
            neighbor_value = objective_function(neighbor)
            print("Checking x =", neighbor, "f(x) =", neighbor_value)

            # Move to the first better neighbor
            if neighbor_value > current_value:
                current = neighbor
                current_value = neighbor_value
                path.append((current, current_value))
                print("Moved to x =", current, "f(x) =", current_value)
                found_better = True
                break

        # Stop if no better neighbor is found
        if not found_better:
            break

    return current, current_value, path


# Main Program
start_state = int(input("Enter starting value of x: "))

best_x, best_value, path = simple_hill_climbing(start_state)

print("\nOptimal solution found:")
print("x =", best_x)
print("f(x) =", best_value)

# ------------------ Graph Plot ------------------

# Generate x values for the graph
x_values = np.linspace(-2, 8, 200)

# Formula to create graph
y_values = objective_function(x_values)

# Plot the function
plt.plot(x_values, y_values, label='f(x) = -(x - 3)^2 + 9')

# Plot hill climbing path
for i, (x, y) in enumerate(path):
    plt.plot(x, y, 'ro')  # red points
    plt.text(x, y + 0.2, f"({x}, {y})")

# Highlight best point
plt.plot(best_x, best_value, 'go', markersize=10, label='Optimal Solution')

# Labels and title
plt.xlabel("x")
plt.ylabel("f(x)")
plt.title("Simple Hill Climbing Graph")
plt.legend()
plt.grid(True)

# Show graph
plt.show()
```
